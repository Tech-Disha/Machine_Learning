Implement Gradient Descent Algorithm to minimize given function

```python
import numpy as np
```

```python
# Define the functions
def f(x):
    return 5 * x**2 + np.exp(x)

def g(x):
    return 5 * x**4 + 7 * x
```

```python
# Generate random dataset
np.random.seed(0)
X = np.random.rand(50)  # 50 random samples between 0 and 1
```

```python
# Calculate Y using the equations
Y_f = f(X)
Y_g = g(X)
```

```python
# Calculate the summation of Y
sum_Y_f = np.sum(Y_f)
sum_Y_g = np.sum(Y_g)
```

```python
# Gradient descent parameters
alpha = 0.01  # Learning rate
iterations = 100
```

```python
# Gradient descent for function f(x)
theta_f = 0  # Initial parameter value for f(x)

for _ in range(iterations):
    df = 10 * theta_f + np.exp(theta_f)  # Derivative of f(x)
    theta_f = theta_f - alpha * df  # Update parameter
```

```python
# Calculate Y using the updated parameter for f(x)
Y_f_updated = f(X)
```

```python
# Gradient descent for function g(x)
theta_g = 0  # Initial parameter value for g(x)

for _ in range(iterations):
    dg = 20 * theta_g**3 + 7  # Derivative of g(x)
    theta_g = theta_g - alpha * dg  # Update parameter
```

```python
# Calculate Y using the updated parameter for g(x)
Y_g_updated = g(X)
```

```python
# Print the results
print("Summation of Y for f(x):", sum_Y_f)
print("Updated parameter for f(x):", theta_f)
```

```python
print("Summation of Y for g(x):", sum_Y_g)
print("Updated parameter for g(x):", theta_g)
```
