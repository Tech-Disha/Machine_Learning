```python
data = pd.read_csv('score_updated.csv')
```

```python
f1 = data['X'].values
f2 = data['Y'].values
```

```python
n = len(data["X"])
```

```python
a = np.mean(f1)
b = np.mean(f2)
x_sumed = np.sum(f1)
y_sumed = np.sum(f2)
```

```python
data["x_minus"] = f1 - a
data["y_minus"] = f2 - b
data["x_y"] = data["x_minus"] * data["y_minus"]
x_y = data["x_y"].sum()
```

```python
data["x_squared"] = data["x_minus"] ** 2
x_square = data["x_squared"].sum()
```

```python
b1 = x_y / x_square
b2 = (1 / n) * (y_sumed - (b1 * x_sumed))
```

```python
j = int(input("enter a value of x you want to calculate : "))
```

```python
i = (b1 * j) + (b2)
```

```python
print(f"value of y when x is {j} is {i}")
```

```python
plt.scatter(f1, f2, label='Data Points', color='blue')
plt.plot(f1, b1 * f1 + b2, color='red', label='Regression Line')
```

```python
plt.xlabel('hours')
plt.ylabel('scores')
plt.legend()
plt.grid()
plt.title('Linear Regression Graph')
plt.show()
```
