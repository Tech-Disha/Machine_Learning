```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('score_updated.csv')
```

```python
X = data['X'].values
Y = data['Y'].values
```

```python
beta0 = 0
beta1 = 1
alpha = 0.001
```

```python
def gradient_descent_step(X, Y, beta0, beta1, alpha):
    n = len(X)
    residuals = Y - (beta0 + beta1 * X)
    residuals1 = X * residuals

    partial_derivative_beta0 = -(2/n) * np.sum(residuals)
    partial_derivative_beta1 = -(2/n) * np.sum(residuals1)

    new_beta0 = beta0 - (alpha * partial_derivative_beta0)
    new_beta1 = beta1 - (alpha * partial_derivative_beta1)

    return new_beta0, new_beta1
```

```python
num_iterations = 50
cost_history = []
```

```python
for i in range(num_iterations):
    new_beta0, new_beta1 = gradient_descent_step(X, Y, beta0, beta1, alpha)
    beta0 = new_beta0
    beta1 = new_beta1
```

```python
new_cost = np.sum((Y - (new_beta0 + new_beta1 * X))**2) / (2 * len(X))
cost_history.append(new_cost)
```

```python
print(f"Final Cost: {cost_history[-1]}")
print(f"Final beta0: {beta0}")
print(f"Final beta1: {beta1}")
```
