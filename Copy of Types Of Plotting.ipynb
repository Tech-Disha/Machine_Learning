```r
#Pie Chart
x <- c(44,12,39,53)
labels <- c("Califoria", "Paris", "Moscow", "Mumbai")
# Plot the chart
pie(x,labels = lbls,col = rainbow(length(lbls)), main = "city_pie_chart")
```

```r
# 3D pie chart
install.packages("plotrix")
library(plotrix)
pie3D(x,labels = lbls,explode = 0.1, main = "city_pie_chart")
```

```r
#Bar chart
H<- c(25,12,43,7,51)
M<- c("Delhi","Beijing", "Tokyo", "Moscow","Goa")
#plot the bar chart
barplot(H, names.arg = M)
```

```r
#Stacked Bar plot with colors and legend
counts<- table(mtcars$vs, mtcars$gear)
counts
barplot(counts,
                main = "Car Distribution by Gears and VS",
                xlab = "Number of Gears",
                col = c("darkblue", "red"),
                legend = rownames(counts))
```

```r
# Box plot
boxplot(mpg~cyl, data = mtcars,xlab = "Number Of Cylinders",
        ylab = "Miles Per Gallon", main = "Mileage Data", col=rainbow(3),
        varwidth=TRUE, notch=TRUE)
```

```r
#Scatter Plot
plot(x=mtcars$wt, y = mtcars$mpg,
     xlab = "Weight",
     ylab = "Mileage",
     xlim = c(2.5,5), ylim = c(15,30), main = "Weight vs Mileage",
     col = "red", pch=20) #pch is for type of Dot.
```

```r
#Scatter plot matrices
pairs(~wt+mpg+disp+cyl,
      data = mtcars, main = "Scatter plot Matrix",
      lower.panel= NULL,pch = 10, col = rainbow(4))
```

```r
#Line Plot
v <- c(8,14,26,5,43)
plot(v,type = "o", col="red",
      xlab = "Month",
      ylab = "Rain Fall",
      main = "Rain Fall Chart"  )
```

```r
#Scatter plot
df_mtcars <- mtcars
ggplot(data=df_mtcars, aes(x=df_mtcars$wt, y = df_mtcars$mpg))+
  geom_point(shape=18,color="dark grey")+
  geom_smooth(method = lm, linetype="dashed", color="red")
```
