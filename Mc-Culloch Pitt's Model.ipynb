Mc-Culloch Pitt's Model

```python
# Fixed inputs x1, x2, x3
x1 = [0, 0, 1, 1, 0, 0, 1, 1]
x2 = [0, 1, 0, 1, 0, 1, 0, 1]
x3 = [1, 1, 1, 1, 0, 0, 0, 0]
```

```python
# Function to check if the condition is satisfied
def check_condition(x, weights, threshold):
    if sum(xi * wi for xi, wi in zip(x, weights)) >= threshold:
        return True
    else:
        return False
```

```python
# Function to take weights and threshold from the user
def get_user_inputs():
    w1 = float(input("Enter weight for w1: "))
    w2 = float(input("Enter weight for w2: "))
    w3 = float(input("Enter weight for w3: "))
    t = float(input("Enter threshold (t): "))

    return [w1, w2, w3], t
```

```python
# Initial user inputs for weights and threshold
weights, threshold = get_user_inputs()

i=0
# Check if the condition is satisfied
while not check_condition([x1[i], x2[i], x3[i]], weights, threshold):
    print("Neuron did not fire for input:", x1[i], x2[i], x3[i], ". Try again.")
    i += 1
    if i == len(x1):
        print("All input combinations tried. Exiting.")
        break
    # Get new weights and threshold from the user
    weights, threshold = get_user_inputs()
```

```python
# If the condition is satisfied, print that the neuron will fire
if i < len(x1):
    print("Yes, the neuron will fire for input:", x1[i], x2[i], x3[i])
```
