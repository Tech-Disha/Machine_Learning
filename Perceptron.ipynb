Perceptron

```python
import numpy as np

def main():
    # Initialize the predefined matrix
    predefined_matrix = np.array([[1], [1], [1], [1]])

    # Get user input for the weights matrix
    weights_matrix = np.array([float(input(f"Enter weight {i+1}: ")) for i in range(4)]).reshape(1, 4)

    # Get user input for the desired output
    desired_output = float(input("Enter the desired output (0 or 1): "))

    # Set learning rate
    learning_rate = 0.1

    # Initialize iteration counter
    iteration = 1

    while True:
        # Multiply weights matrix with predefined matrix
        final_output = np.dot(weights_matrix, predefined_matrix)

        # Apply the activation function (0 if <= 0, 1 otherwise)
        final_output = 1 if final_output > 0 else 0

        # Print current weights and final output
        print(f"Iteration {iteration}:")
        print("Weights Matrix:", weights_matrix)
        print("Final Output:", final_output)

        # Check if the final output matches the desired output
        if final_output == desired_output:
            print("Success! The final output matches the desired output.")
            break

        # Update weights matrix
        weights_matrix += learning_rate * (desired_output - final_output) * predefined_matrix.T

        # Print updated weights matrix
        print("Updated Weights Matrix:", weights_matrix)
        print()

        iteration += 1

if __name__ == "__main__":
    main()
```
